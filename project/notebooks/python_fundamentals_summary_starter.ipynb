# Reusable Utility Functions


## 18. Utility Function — Summary Statistics

Write a reusable function called:

```python
get_summary_stats(df)
```

It should:
1. Select numeric columns.
2. Return useful summary statistics for those columns.

You may start with `df.describe()`, or build something more customized.


````python
# TODO: Write get_summary_stats(df)
def get_summary_stats(df):
    summary = {
        'mean': df['amount'].mean(),
        'median': df['amount'].median(),
        'std': df['amount'].std(),
        'min': df['amount'].min(),
        'max': df['amount'].max(),
        'sum': df['amount'].sum()
    }
    return summary


# TODO: Test it on df
get_summary_stats(df)
````

## 19. Utility Function — Column Name Cleaner

Write a function:

```python
clean_column_names(df)
```

It should return a copy of the DataFrame where column names:
- are lowercase
- have leading/trailing spaces removed
- replace spaces with underscores

Example:

```text
"Transaction ID " -> "transaction_id"
```


````python
# TODO: Write clean_column_names(df)
def clean_column_names(df):
    df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_').str.replace('(', '').str.replace(')', '')
    return df


# TODO: Test it with a small example DataFrame
test_df = pd.DataFrame({
    ' Column 1 ': [1, 2, 3],
    ' Column 2 ': [4, 5, 6]
})
print("Before cleaning:")
print(test_df)
clean_column_names(test_df)
````
