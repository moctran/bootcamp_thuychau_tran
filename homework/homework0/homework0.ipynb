## Framing Worksheet

### Problem Statement
A large transaction dataset contains information not only about individual account activity but also about relationships between accounts. This project models transactions as a network so that structural patterns can be converted into reusable graph features for customer, risk, and anomaly analytics.

### Stakeholder & User
- **Decision owner:** Data Science / Risk / Customer Analytics lead
- **Tool/operator:** Data scientists and analysts
- **Workflow:** Recurring transaction snapshots, such as monthly processing

### Useful Answer
- **Descriptive:** Explain account position and relationship structure
- **Predictive:** Determine whether network features improve downstream models
- **Artifact:** Scalable feature pipeline + processed feature tables + evaluation report

### Assumptions & Constraints
- Sender/receiver identifiers, amount, and timestamp are available
- Graph size can be very large
- Scalable GPU/distributed processing may be needed
- Customer identifiers and sensitive data must be protected
- Predictive features must respect time boundaries

### Known Unknowns / Risks
- Hub accounts can distort graph metrics
- Missing transactions can alter topology
- Feature stability may vary over time
- Computation may be expensive
- Temporal leakage may falsely improve predictive performance

### Lifecycle Mapping
- Scope problem → Stage 01 → README + stakeholder artifact
- Prepare data → clean edge list
- Build graph → directed/weighted graph
- Engineer features → node/edge feature tables
- Evaluate → stability, scalability, predictive lift
- Deliver → reusable scripts + documentation


````python
from pathlib import Path
import textwrap

# Change this if you prefer another repository name.
REPO_NAME = "transaction_network_project"

# If this notebook is saved inside homework/homework0, the project
# will be created there.
base = Path.cwd() / REPO_NAME

FOLDERS = [
    "data/raw",
    "data/processed",
    "src",
    "notebooks",
    "docs",
    "reports",
    "model",
]

base.mkdir(parents=True, exist_ok=True)
for folder in FOLDERS:
    (base / folder).mkdir(parents=True, exist_ok=True)

print(f'Project directory created at: {base.resolve()}')
````

````python
README_TEMPLATE = '# Transaction Network Analysis\n\n**Stage:** Problem Framing & Scoping (Stage 01)\n\n## Problem Statement\nA large retail bank processes millions of customer transactions each month. Looking at transactions only as isolated rows can miss important relationship patterns between accounts. By representing accounts as nodes and transactions as edges, network analysis can reveal structural behaviors such as highly connected accounts, repeated reciprocal flows, concentrated counterparties, and tightly connected neighborhoods.\n\nThe project aims to build a scalable transaction-network analytics pipeline that converts raw transaction records into account-level and relationship-level graph features. These features can support downstream analytical tasks such as customer behavior analysis, anomaly detection, churn modeling, and credit-risk modeling.\n\n## Stakeholder & User\n- **Decision owner:** Data Science / Risk / Customer Analytics lead.\n- **Tool/operator:** Data scientists and analysts working with large-scale transaction data.\n- **Workflow context:** The pipeline is run on recurring transaction snapshots, such as monthly data, and produces reusable network features for downstream models and analysis.\n- **Decision supported:** Determine which network-derived signals are useful enough to include in customer analytics or predictive models.\n\n## Useful Answer & Decision\n- **Type:** Primarily descriptive and predictive.\n- **Descriptive goal:** Summarize how each account is positioned and behaves within the transaction network.\n- **Predictive goal:** Test whether graph-derived features improve downstream prediction compared with using conventional customer and transaction features alone.\n- **Core artifacts:** A reproducible graph-feature pipeline, processed feature tables, exploratory analysis, and model-ready datasets.\n- **Example features:** Degree, weighted degree, reciprocity, neighbor entropy, neighborhood overlap, and other node/edge statistics.\n- **Evaluation:** Feature stability, computational scalability, data quality, and incremental predictive performance in downstream models.\n\n## Assumptions & Constraints\n- Transaction records contain reliable sender, receiver, amount, and time information.\n- Account identifiers can be consistently linked across the analysis period.\n- The graph may contain millions of transaction records, so single-machine graph processing may be too slow or memory intensive.\n- GPU/distributed tools such as cuGraph and Dask may be required for scalable computation.\n- Transaction direction and repeated transactions carry useful behavioral information and should be preserved where appropriate.\n- Sensitive customer information should not be exposed in outputs; analysis should rely on internal identifiers and approved data access.\n- Network features should be calculated without leaking future information into predictive models.\n\n## Known Unknowns / Risks\n- Some network patterns may reflect normal high-volume activity rather than meaningful behavioral signals.\n- Very large merchants or hub accounts may dominate degree-based metrics.\n- Missing or filtered transactions may distort local network structure.\n- Network features may change substantially across months, reducing stability.\n- Feature computation may become expensive as graph size grows.\n- Apparent predictive gains may come from temporal leakage if feature windows are not aligned correctly.\n- The most useful graph features may differ across churn, credit-risk, and anomaly-detection tasks.\n\n## Lifecycle Mapping\nGoal → Stage → Deliverable\n\n- Define the business and analytics question → **Problem Framing & Scoping (Stage 01)** → Project scope and README.\n- Ingest and validate transaction data → **Data Preparation** → Clean edge list and account mapping.\n- Build the transaction graph → **Network Construction** → Directed/weighted graph representation.\n- Engineer graph features → **Feature Engineering** → Node- and edge-level feature tables.\n- Explore network behavior → **EDA** → Summary statistics and visual diagnostics.\n- Test downstream value → **Modeling & Evaluation** → Baseline-vs-network-feature comparison.\n- Package reusable workflow → **Delivery** → Reproducible scripts, documentation, and reports.\n\n## Repo Plan\n- `data/raw/` — original or sample transaction inputs; sensitive production data should not be committed.\n- `data/processed/` — cleaned edge lists and generated feature tables.\n- `src/` — reusable graph construction and feature-engineering code.\n- `notebooks/` — exploratory analysis and experiments.\n- `docs/` — stakeholder notes, scope, data dictionary, and methodology.\n- `reports/` — figures, summaries, and evaluation results.\n- `model/` — saved downstream models or model metadata when applicable.\n\n## Initial Technical Workflow\n1. Load a monthly transaction snapshot.\n2. Clean identifiers, timestamps, amounts, and transaction direction.\n3. Aggregate records into an edge list where appropriate.\n4. Construct a directed and/or weighted transaction graph.\n5. Compute scalable graph features.\n6. Join graph features back to account-level analytical datasets.\n7. Evaluate feature distributions, stability, and usefulness.\n8. Compare downstream models with and without network features.\n\n## Success Criteria\nThe project is successful if it produces a reproducible and scalable network-feature pipeline and demonstrates which graph features provide stable, interpretable, and useful signals for downstream customer or risk analytics.'

readme_path = base / "README.md"
readme_path.write_text(README_TEMPLATE, encoding='utf-8')
print(f'Created: {readme_path}')
````

````python
STAKEHOLDER_ARTIFACT = '# Stakeholder Artifact — Transaction Network Analytics\n\n## Stakeholder\nData Science / Risk / Customer Analytics lead\n\n## Need\nThe stakeholder needs a scalable way to extract relationship-based signals from large transaction datasets rather than relying only on account-level aggregates.\n\n## Decision\nDecide whether network-derived features should be incorporated into recurring analytics and predictive models.\n\n## Output\nA reproducible transaction-network pipeline that produces documented account-level graph features and an evaluation of their incremental value.\n\n## Key Questions\n- Which accounts occupy unusual or influential positions in the network?\n- Which features are stable across monthly snapshots?\n- Which network signals add information beyond traditional transaction features?\n- Can the pipeline scale to millions of transaction records?\n- Are the features interpretable and safe from temporal leakage?'

stakeholder_path = base / "docs" / "stakeholder_artifact.md"
stakeholder_path.write_text(STAKEHOLDER_ARTIFACT, encoding='utf-8')
print(f'Created: {stakeholder_path}')
````

````python
# Add placeholder files so empty folders can be tracked by Git.
for folder in ["data/raw", "data/processed", "src", "notebooks", "reports", "model"]:
    keep = base / folder / ".gitkeep"
    keep.touch(exist_ok=True)

# Protect common sensitive/local files from Git.
gitignore = textwrap.dedent('''
.env
.venv/
__pycache__/
*.pyc
.DS_Store

# Do not commit raw/processed transaction data
data/raw/*
data/processed/*
!data/raw/.gitkeep
!data/processed/.gitkeep
''').strip() + '\n'

(base / ".gitignore").write_text(gitignore, encoding="utf-8")
print('Created .gitignore and .gitkeep files.')
````

````python
# Show the generated project tree.
for path in sorted(base.rglob('*')):
    relative = path.relative_to(base)
    indent = '    ' * (len(relative.parts) - 1)
    marker = '📁' if path.is_dir() else '📄'
    print(f'{indent}{marker} {relative.name}')
````
