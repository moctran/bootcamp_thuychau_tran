# Stage 01 Homework — Problem Framing & Scoping

## Transaction Network Analysis for Risk Management


## 1. Project Scoping Paragraph

Financial institutions process large volumes of transactions every day, making it difficult for risk teams to understand how accounts are connected and where potentially unusual patterns may be concentrated. Looking at transactions one by one can miss broader structures, such as accounts that interact with unusually many counterparties, tightly connected groups of accounts, repeated two-way transfers, or concentrated flows through a small number of entities. This project will represent transaction activity as a **network**, where accounts are nodes and transactions between them form edges, and use descriptive network analysis to summarize these structures.

The primary stakeholder is the **Transaction Risk Management team**, which is responsible for identifying areas that may require further review and deciding how investigative resources should be prioritized. The day-to-day users may be risk analysts or investigators, while data and analytics teams support the project technically. The goal is **descriptive rather than predictive or causal**: the project will not attempt to prove that an account is fraudulent or predict future fraud. Instead, it will produce interpretable network indicators and visual summaries that help the risk team understand transaction relationships, identify unusual network structures, and prioritize accounts or groups for deeper investigation.


## 2. Stakeholder & User

### Decision owner
**Transaction Risk Management / Financial Crime Risk Management team**

The risk team decides:
- which accounts, relationships, or clusters deserve further review;
- how to prioritize limited investigation capacity;
- whether an observed pattern should be escalated for additional checks.

### Primary day-to-day users
**Risk analysts / transaction investigators**

They use the output to:
- understand how an account is connected to the rest of the transaction network;
- compare an account's activity with typical network behavior;
- inspect groups of strongly connected accounts;
- support investigation with interpretable evidence.

### Supporting technical users
**Data analysts / data scientists / data engineers**

Their role is to:
- prepare transaction data;
- construct the network;
- calculate network features;
- maintain reproducible analysis.

The technical team supports the analysis, but it is **not the main stakeholder**, because the final purpose of the project is to improve a business risk decision rather than to optimize a technical system.


## 3. Useful Answer & Decision

### Type of answer
**Descriptive**

The project asks:

> **What structures and relationships exist in the transaction network, and which accounts or groups appear unusual enough to deserve additional review?**

This is descriptive because the analysis summarizes observed transaction behavior. It does **not** claim that unusual network behavior causes fraud, nor does it attempt to predict whether an account will commit fraud in the future.

### Useful outputs

Examples of useful descriptive indicators include:

- number of unique counterparties;
- incoming and outgoing transaction relationships;
- transaction volume or value concentration;
- reciprocity of relationships;
- repeated or dense account-to-account interactions;
- local clustering or neighborhood structure;
- connected components or communities;
- accounts that act as hubs or bridges between groups.

### Business artifact

A useful deliverable would be a **risk-oriented network summary/dashboard or investigation-priority report** containing:

1. a ranked or filtered view of accounts with unusual network characteristics;
2. interpretable network metrics explaining why each account is noteworthy;
3. summaries of important account groups or communities;
4. selected network visualizations for investigation;
5. clear caveats that an unusual network pattern is a **screening signal, not proof of misconduct**.

### Decision supported

The output should help the risk team answer:

> **Which accounts or groups should we examine first, and what network evidence makes them worth examining?**


## 4. Assumptions & Constraints

### Assumptions

- Each transaction can be linked to a sender and a receiver.
- Account identifiers can be represented consistently across the observation period.
- Transaction relationships contain useful information beyond individual transaction-level attributes.
- Historical transaction data is sufficiently complete to describe the network during the chosen period.
- Unusual network structure can be useful for **screening**, even though it does not by itself imply fraudulent behavior.

### Constraints

- Transaction data may be extremely large, so the full network may not fit comfortably into memory on a single machine.
- Customer and transaction data is sensitive and must be handled under privacy, security, and compliance requirements.
- Account identifiers should be anonymized or pseudonymized for analysis.
- Network metrics may be expensive to compute on very large graphs.
- The analysis may need to operate on a defined time window, such as a week or month.
- Risk teams need interpretable outputs; highly complex metrics are less useful if investigators cannot understand why an account was surfaced.
- Ground-truth fraud labels may be unavailable or incomplete, which is one reason this stage focuses on descriptive analysis.


## 5. Known Unknowns / Risks

- **What counts as unusual?** High connectivity may be completely normal for merchants, payment processors, or other high-volume customers.
- **Customer type matters.** Network metrics may need to be interpreted relative to peer groups rather than with one threshold for every account.
- **Time-window sensitivity.** A network built over one day, one month, or one year can produce very different structures.
- **Missing external relationships.** Transactions outside the institution may not be visible, so the observed graph is only a partial network.
- **False positives.** Unusual network structure does not automatically indicate financial crime.
- **Changing behavior.** Normal transaction behavior can evolve over time, so descriptive benchmarks may need periodic updates.
- **Investigation capacity.** Even a useful analysis must produce a manageable number of cases for the risk team.

### How these risks could be tested later

Future stages could:
- compare network metrics across customer segments;
- test several observation windows;
- compare surfaced accounts with historical investigation outcomes where available;
- collect investigator feedback on whether the indicators are understandable and useful;
- monitor how many accounts are surfaced relative to investigation capacity.
